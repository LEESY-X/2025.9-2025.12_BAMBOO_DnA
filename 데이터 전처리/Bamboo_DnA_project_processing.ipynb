```python
#데이터를 병합하는 것 보다 3개의 파일로 관리하는 것이 적절하다고 판단
df_lg = pd.DataFrame(file_lg)
df_samsung = pd.DataFrame(file_samsung)
df_sk = pd.DataFrame(file_sk)
```

```python
# 원하는 속성만 조회 및 저장
df_lg1 = df_lg[['일자', '종가', '대비', '등락률', '거래량']]
df_lg1.columns = ['Date', '종가', '차액', '등락률', '거래량']
df_lg1.sort_values(by='Date', ascending=True, inplace=True)
df_lg1=df_lg1.reset_index(drop=True)

df_samsung1 = df_samsung[['일자', '종가', '대비', '등락률', '거래량']]
df_samsung1.columns = ['Date', '종가', '차액', '등락률', '거래량']
df_samsung1.sort_values(by='Date', ascending=True, inplace=True)
df_samsung1=df_samsung1.reset_index(drop=True)

df_sk1 = df_sk[['일자', '종가', '대비', '등락률', '거래량']]
df_sk1.columns = ['Date', '종가', '차액', '등락률', '거래량']
df_sk1.sort_values(by='Date', ascending=True, inplace=True)
df_sk1=df_sk1.reset_index(drop=True)

df_lg1.to_csv('/content/drive/MyDrive/Colab Notebooks/2025-2/[25.9~25.12] Bamboo DnA/DnA_project (Stock_and_News)/Data/LG_energysolution/Stocks/2025_processing_lg.csv', encoding="cp949")
df_samsung1.to_csv('/content/drive/MyDrive/Colab Notebooks/2025-2/[25.9~25.12] Bamboo DnA/DnA_project (Stock_and_News)/Data/SAMSUNG/Stocks/2025_processing_samsung.csv', encoding="cp949")
df_sk1.to_csv('/content/drive/MyDrive/Colab Notebooks/2025-2/[25.9~25.12] Bamboo DnA/DnA_project (Stock_and_News)/Data/SK_hynix/Stocks/2025_processing_sk.csv', encoding="cp949")
```
